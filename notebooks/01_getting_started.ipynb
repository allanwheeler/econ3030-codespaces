## C. Importing data

A CSV is a plain text table. The path below goes up from the notebook folder and into `data`. Each row records one state in one year. Columns contain variables: state, year, and unemployment rate. A value of `5.2` means 5.2%, not 520%.

This fixed synthetic dataset has five states and ten years, 2014 through 2023. It was created for this lab, with a noticeable increase in 2020. It cannot support claims about actual state economies.

```python
df = pl.read_csv("../data/state_unemployment.csv")

df.head()
```

## D. Filtering data

Filtering keeps rows that meet a condition. Here we keep rows whose state is California. The double equals sign asks whether two values match; the single equals sign gives a result a name.

```python
california = df.filter(
    pl.col("state") == "California"
)

california
```

Check that all ten rows say California. Which years seem unusually high?

## E. Summarizing data

Aggregation combines observations into a summary. For example, the mean of rates 4 and 6 is 5. Here we group the rows by state, average the ten annual rates in each group, and sort from highest to lowest.

This is a simple average of annual rates, not a population weighted national unemployment rate. `alias` gives the calculated column a readable name.

```python
summary = (
    df.group_by("state")
      .agg(
          pl.col("unemployment_rate")
            .mean()
            .alias("average_unemployment")
      )
      .sort("average_unemployment", descending=True)
)

summary
```

## F. Visualization

A line chart shows change over time. Altair uses the Polars table directly. `year:O` puts years in order; `unemployment_rate:Q` treats rates as numbers.

The axis adds a percent sign to the stored value. A percent formatter would multiply 5.2 by 100, so we do not use it here. Hover over a point to see the rate.

```python
chart = (
    alt.Chart(california)
    .mark_line(point=True)
    .encode(
        x=alt.X("year:O", title="Year", sort="ascending"),
        y=alt.Y("unemployment_rate:Q", title="Unemployment rate (%)",
                scale=alt.Scale(zero=True),
                axis=alt.Axis(labelExpr="datum.value + '%'")),
        tooltip=[alt.Tooltip("year:O", title="Year"),
                 alt.Tooltip("unemployment_rate:Q", title="Unemployment rate (%)", format=".1f")]
    )
    .properties(title="California unemployment, 2014 to 2023 (synthetic)", width=520, height=280)
)

chart
```
